## <center>**Difusi Neutron 2 Dimensi Silinder (r, z) Multigrup**</center> 

- **Grup I**
$$ D_1 \nabla^2 \phi^1_{(r,z)} - \Sigma_{a1} \phi^1_{(r,z)} - \Sigma_{s1 \rightarrow 2} \phi^1_{(r,z)} - \Sigma_{s1 \rightarrow 3} \phi^1_{(r,z)} - \Sigma_{s1 \rightarrow 4} \phi^1_{(r,z)}- \Sigma_{s1 \rightarrow 5} \phi^1_{(r,z)} + \frac{\chi_1}{k} (\nu_1 \Sigma_{f1} \phi^1_{(r,z)} +...+ \nu_5 \Sigma_{f5} \phi^5_{(r,z)}) = 0$$
- **Grup II**
$$ D_2 \nabla^2 \phi^2_{(r,z)} - \Sigma_{a2} \phi^2_{(r,z)} - \Sigma_{s2 \rightarrow 3} \phi^2_{(r,z)} - \Sigma_{s2 \rightarrow 4} \phi^2_{(r,z)} - \Sigma_{s2 \rightarrow 5} \phi^2_{(r,z)} + \Sigma_{s1 \rightarrow 2} \phi^1_{(r,z)} + \frac{\chi_2}{k} (\nu_1 \Sigma_{f1} \phi^1_{(r,z)} +...+ \nu_5 \Sigma_{f5} \phi^5_{(r,z)}) = 0$$
- **Grup III**
$$ D_3 \nabla^2 \phi^3_{(r,z)} - \Sigma_{a3} \phi^3_{(r,z)} - \Sigma_{s3 \rightarrow 4} \phi^3_{(r,z)} - \Sigma_{s3 \rightarrow 5} \phi^3_{(r,z)} + \Sigma_{s1 \rightarrow 3} \phi^1_{(r,z)} + \Sigma_{s2 \rightarrow 3} \phi^2_{(r,z)} + \frac{\chi_3}{k} (\nu_1 \Sigma_{f1} \phi^1_{(r,z)} +...+ \nu_5 \Sigma_{f5} \phi^5_{(r,z)})= 0$$
- **Grup IV**
$$ D_4 \nabla^2 \phi^4_{(r,z)} - \Sigma_{a4} \phi^4_{(r,z)} - \Sigma_{s4 \rightarrow 5} \phi^4_{(r,z)} + \Sigma_{s1 \rightarrow 4} \phi^1_{(r,z)} + \Sigma_{s2 \rightarrow 4} \phi^2_{(r,z)}+ \Sigma_{s3 \rightarrow 4} \phi^3_{(r,z)} + \frac{\chi_4}{k} (\nu_1 \Sigma_{f1} \phi^1_{(r,z)} +...+ \nu_5 \Sigma_{f5} \phi^5_{(r,z)})= 0$$
- **Grup V**
$$ D_5 \nabla^2 \phi^5_{(r,z)} - \Sigma_{a5} \phi^5_{(r,z)} + \Sigma_{s1 \rightarrow 5} \phi^1_{(r,z)} + \Sigma_{s2 \rightarrow 5} \phi^2_{(r,z)} + \Sigma_{s3 \rightarrow 5} \phi^3_{(r,z)}+\Sigma_{s4 \rightarrow 5} \phi^4_{(r,z)} + \frac{\chi_5}{k} (\nu_1 \Sigma_{f1} \phi^1_{(r,z)} +...+ \nu_5 \Sigma_{f5} \phi^5_{(r,z)}) = 0$$

**Laplacian silinder arah (r,z):**
$$ \nabla^2 f_{(r,z)} = \frac{1}{r} \frac{\partial}{\partial r} (r \frac{\partial f_{(r,z)}}{\partial r})+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$
$$ \nabla^2 f_{(r,z)} = \frac{1}{r} (r \frac{\partial}{\partial r} \frac{\partial f_{(r,z)}}{\partial r}+ \frac{\partial r}{\partial r} \frac{\partial f_{(r,z)}}{\partial r})+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$
$$ \nabla^2 f_{(r,z)} = \frac{1}{r} (r \frac{\partial^2 f_{(r,z)}}{\partial r^2}+ 1 \frac{\partial f_{(r,z)}}{\partial r})+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$
$$ \nabla^2 f_{(r,z)} =  \frac{\partial^2 f_{(r,z)}}{\partial r^2}+ \frac{1}{r} \frac{\partial f_{(r,z)}}{\partial r}+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$

Diskritisasi arah r, dengan menggunakan central difference
$$ \frac{\partial f_{(r,z)}}{\partial r} = \frac{f_{(r+\Delta r, z)}- f_{(r-\Delta r,z)}}{2 \Delta r} $$
$$ \frac{\partial^2 f_{(r,z)}}{\partial r^2} = \frac{f_{(r+\Delta r, z)}-2 f_{(r,z)}+ f_{(r-\Delta r,z)}}{\Delta r^2} $$
Diskritisasi arah z, dengan menggunakan central difference
$$ \frac{\partial^2 f_{(r,z)}}{\partial z^2} = \frac{f_{(r, z+\Delta z)}-2 f_{(r,z)}+ f_{(r,z-\Delta z)}}{\Delta z^2} $$

Maka setelah di diskritisasi persamaan menjadi
**$$ \nabla^2 f_{(r,z)} = \frac{f_{(r+\Delta r, z)}-2 f_{(r,z)}+ f_{(r-\Delta r,z)}}{\Delta r^2} + \frac{1}{r} \frac{f_{(r+\Delta r, z)}- f_{(r-\Delta r,z)}}{2 \Delta r} + \frac{f_{(r, z+\Delta z)}-2 f_{(r,z)}+ f_{(r,z-\Delta z)}}{\Delta z^2} $$**

**Substitusi hasil diskritisasi laplacian ke dalam persamaan difusi grup $g$**

Untuk titik interior ($i \geq 1$, dengan $r_i = i\,\Delta r$), $g = 1,\dots,5$ (hanya *down-scattering*, $g' < g$):

$$ D_g \left(\frac{\phi^g_{i+1,j}-2 \phi^g_{i,j}+ \phi^g_{i-1,j}}{\Delta r^2} + \frac{1}{r_i} \frac{\phi^g_{i+1,j}- \phi^g_{i-1,j}}{2 \Delta r} + \frac{\phi^g_{i,j+1}-2 \phi^g_{i,j}+ \phi^g_{i,j-1}}{\Delta z^2}\right) - \Sigma_{R,g}\,\phi^g_{i,j} + \sum_{g'<g} \Sigma_{s,g' \rightarrow g}\,\phi^{g'}_{i,j} + \frac{\chi_g}{k}\sum_{g'=1}^{5} \nu_{g'}\Sigma_{f,g'}\,\phi^{g'}_{i,j} = 0 $$

dengan penampang lintang *removal* (absorpsi + hamburan keluar grup):

$$ \Sigma_{R,1} = \Sigma_{a1}+\Sigma_{s1 \rightarrow 2}+\Sigma_{s1 \rightarrow 3}+\Sigma_{s1 \rightarrow 4}+\Sigma_{s1 \rightarrow 5} \qquad \Sigma_{R,2} = \Sigma_{a2}+\Sigma_{s2 \rightarrow 3}+\Sigma_{s2 \rightarrow 4}+\Sigma_{s2 \rightarrow 5} $$
$$ \Sigma_{R,3} = \Sigma_{a3}+\Sigma_{s3 \rightarrow 4}+\Sigma_{s3 \rightarrow 5} \qquad \Sigma_{R,4} = \Sigma_{a4}+\Sigma_{s4 \rightarrow 5} \qquad \Sigma_{R,5} = \Sigma_{a5} $$

**Dikelompokkan terhadap $\phi$ (titik interior):**

$$ a^g_W\,\phi^g_{i-1,j} + a^g_E\,\phi^g_{i+1,j} + a^g_z\left(\phi^g_{i,j-1}+\phi^g_{i,j+1}\right) - a^g_P\,\phi^g_{i,j} + \sum_{g'<g} \Sigma_{s,g' \rightarrow g}\,\phi^{g'}_{i,j} + \frac{\chi_g}{k}\sum_{g'=1}^{5} \nu_{g'}\Sigma_{f,g'}\,\phi^{g'}_{i,j} = 0 $$

dengan koefisien (indeks $g$ menunjukkan grup, **bukan** huruf $A,B,C,D,E$ agar tidak bentrok dengan konstanta difusi $D_g$):

$$ a^g_W = \frac{D_g}{\Delta r^2}-\frac{D_g}{2 r_i \Delta r} \qquad a^g_E = \frac{D_g}{\Delta r^2}+\frac{D_g}{2 r_i \Delta r} \qquad a^g_z = \frac{D_g}{\Delta z^2} \qquad a^g_P = \frac{2D_g}{\Delta r^2}+\frac{2D_g}{\Delta z^2}+\Sigma_{R,g} $$

Perhatikan $a^g_W \neq a^g_E$ (tanda suku $\frac{1}{r}\frac{\partial}{\partial r}$ berbeda) dan $r_i$ adalah koordinat titik yang ditinjau, bukan jari-jari teras $R$.

**Persamaan tiap grup pada titik interior** ($\mathcal{F}_{i,j} = \nu_1\Sigma_{f1}\phi^1_{i,j}+\dots+\nu_5\Sigma_{f5}\phi^5_{i,j}$):

- **Grup I**
$$ a^1_W\phi^1_{i-1,j} + a^1_E\phi^1_{i+1,j} + a^1_z(\phi^1_{i,j-1}+\phi^1_{i,j+1}) - a^1_P\phi^1_{i,j} + \frac{\chi_1}{k}\mathcal{F}_{i,j} = 0 $$
- **Grup II**
$$ \Sigma_{s1 \rightarrow 2}\phi^1_{i,j} + a^2_W\phi^2_{i-1,j} + a^2_E\phi^2_{i+1,j} + a^2_z(\phi^2_{i,j-1}+\phi^2_{i,j+1}) - a^2_P\phi^2_{i,j} + \frac{\chi_2}{k}\mathcal{F}_{i,j} = 0 $$
- **Grup III**
$$ \Sigma_{s1 \rightarrow 3}\phi^1_{i,j} + \Sigma_{s2 \rightarrow 3}\phi^2_{i,j} + a^3_W\phi^3_{i-1,j} + a^3_E\phi^3_{i+1,j} + a^3_z(\phi^3_{i,j-1}+\phi^3_{i,j+1}) - a^3_P\phi^3_{i,j} + \frac{\chi_3}{k}\mathcal{F}_{i,j} = 0 $$
- **Grup IV**
$$ \Sigma_{s1 \rightarrow 4}\phi^1_{i,j} + \Sigma_{s2 \rightarrow 4}\phi^2_{i,j} + \Sigma_{s3 \rightarrow 4}\phi^3_{i,j} + a^4_W\phi^4_{i-1,j} + a^4_E\phi^4_{i+1,j} + a^4_z(\phi^4_{i,j-1}+\phi^4_{i,j+1}) - a^4_P\phi^4_{i,j} + \frac{\chi_4}{k}\mathcal{F}_{i,j} = 0 $$
- **Grup V**
$$ \Sigma_{s1 \rightarrow 5}\phi^1_{i,j} + \Sigma_{s2 \rightarrow 5}\phi^2_{i,j} + \Sigma_{s3 \rightarrow 5}\phi^3_{i,j} + \Sigma_{s4 \rightarrow 5}\phi^4_{i,j} + a^5_W\phi^5_{i-1,j} + a^5_E\phi^5_{i+1,j} + a^5_z(\phi^5_{i,j-1}+\phi^5_{i,j+1}) - a^5_P\phi^5_{i,j} + \frac{\chi_5}{k}\mathcal{F}_{i,j} = 0 $$

**Kondisi batas**

- **Sumbu $r=0$ (simetri)**: $\partial\phi/\partial r = 0$, sehingga suku $\frac{1}{r}\frac{\partial \phi}{\partial r}$ bernilai $\frac{0}{0}$. Dengan aturan L'Hôpital:
$$ \lim_{r \to 0}\frac{1}{r}\frac{\partial \phi}{\partial r} = \frac{\partial^2 \phi}{\partial r^2} \quad\Rightarrow\quad \nabla^2\phi\Big|_{r=0} = 2\frac{\partial^2 \phi}{\partial r^2}+\frac{\partial^2 \phi}{\partial z^2} $$
Dengan titik hantu $\phi_{-1,j}=\phi_{1,j}$ (simetri) maka $\frac{\partial^2 \phi}{\partial r^2}\big|_{r=0} = \frac{2(\phi_{1,j}-\phi_{0,j})}{\Delta r^2}$, sehingga koefisien di $i=0$ menjadi
$$ a^g_{E,0} = \frac{4D_g}{\Delta r^2} \qquad a^g_{W,0} = 0 \qquad a^g_z = \frac{D_g}{\Delta z^2} \qquad a^g_{P,0} = \frac{4D_g}{\Delta r^2}+\frac{2D_g}{\Delta z^2}+\Sigma_{R,g} $$
- **Permukaan luar** ($r=R$, $z=0$, $z=H$): fluks nol $\phi^g = 0$. Baris matriks $\mathbf{M}$ untuk titik ini diisi identitas dan baris matriks $\mathbf{F}$ diisi nol, sehingga $\phi^g = 0$.

**Satu matriks untuk seluruh grup energi**

Setiap grup memiliki $N = N_r N_z$ titik. Titik $(i,j)$ grup $g$ ($g=1,\dots,5$) diberi indeks global
$$ k_{\text{global}} = \underbrace{i + j N_r}_{p} + (g-1)\,N $$
sehingga vektor fluks seluruh grup tersusun sebagai $\boldsymbol{\Phi} = \left(\boldsymbol{\phi}^{1\,T}, \boldsymbol{\phi}^{2\,T}, \dots, \boldsymbol{\phi}^{5\,T}\right)^T$ berukuran $5N$, dan sistemnya menjadi **satu** persamaan matriks berukuran $5N \times 5N$:

$$
\underbrace{\begin{pmatrix}
L_1 & 0 & 0 & 0 & 0 \\
S_{21} & L_2 & 0 & 0 & 0 \\
S_{31} & S_{32} & L_3 & 0 & 0 \\
S_{41} & S_{42} & S_{43} & L_4 & 0 \\
S_{51} & S_{52} & S_{53} & S_{54} & L_5
\end{pmatrix}}_{\mathbf{M}}
\begin{pmatrix} \boldsymbol{\phi}^1 \\ \boldsymbol{\phi}^2 \\ \boldsymbol{\phi}^3 \\ \boldsymbol{\phi}^4 \\ \boldsymbol{\phi}^5 \end{pmatrix}
= -\frac{1}{k}
\underbrace{\begin{pmatrix}
\chi_1 \nu_1\Sigma_{f1} I & \chi_1 \nu_2\Sigma_{f2} I & \cdots & \chi_1 \nu_5\Sigma_{f5} I \\
\chi_2 \nu_1\Sigma_{f1} I & \chi_2 \nu_2\Sigma_{f2} I & \cdots & \chi_2 \nu_5\Sigma_{f5} I \\
\vdots & \vdots & \ddots & \vdots \\
\chi_5 \nu_1\Sigma_{f1} I & \chi_5 \nu_2\Sigma_{f2} I & \cdots & \chi_5 \nu_5\Sigma_{f5} I
\end{pmatrix}}_{\mathbf{F}}
\begin{pmatrix} \boldsymbol{\phi}^1 \\ \boldsymbol{\phi}^2 \\ \boldsymbol{\phi}^3 \\ \boldsymbol{\phi}^4 \\ \boldsymbol{\phi}^5 \end{pmatrix}
$$

- **Blok diagonal** $L_g$ ($N\times N$): operator difusi $-$ removal. Baris titik interior $p$ berisi
$$ (L_g\boldsymbol{\phi}^g)_p = a^g_W\phi^g_{p-1} + a^g_E\phi^g_{p+1} + a^g_z\phi^g_{p-N_r} + a^g_z\phi^g_{p+N_r} - a^g_P\phi^g_p $$
- **Blok bawah** $S_{gg'} = \Sigma_{s,g' \rightarrow g}\, I_{\text{interior}}$ untuk $g' < g$ (matriks diagonal, hamburan dari grup $g'$ ke $g$). Blok di atas diagonal nol karena tidak ada *up-scattering*.
- **Matriks fisi** $\mathbf{F}$: blok $(g,g')$ adalah $\chi_g\,\nu_{g'}\Sigma_{f,g'}\,I_{\text{interior}}$, atau ditulis $\mathbf{F} = \left(\boldsymbol{\chi}\,(\nu\Sigma_f)^T\right) \otimes I_N$. Seluruh baris grup (bukan hanya grup 1) terisi karena spektrum fisi $\chi_g$ berlaku untuk tiap grup.

**Persamaan nilai eigen dan iterasi sumber (*power iteration*)**

$$ \mathbf{M}\boldsymbol{\Phi} = -\frac{1}{k}\,\mathbf{F}\boldsymbol{\Phi} \quad\Longleftrightarrow\quad (-\mathbf{M})\,\boldsymbol{\Phi} = \frac{1}{k}\,\mathbf{F}\boldsymbol{\Phi} $$

$$ (-\mathbf{M})\,\boldsymbol{\Phi}^{(n+1)} = \frac{1}{k^{(n)}}\,\mathbf{F}\boldsymbol{\Phi}^{(n)} \qquad k^{(n+1)} = k^{(n)}\,\frac{\sum \mathbf{F}\boldsymbol{\Phi}^{(n+1)}}{\sum \mathbf{F}\boldsymbol{\Phi}^{(n)}} $$

Iterasi berhenti ketika perubahan $k$ dan $\boldsymbol{\Phi}$ lebih kecil dari toleransi. $k$ terbesar yang konvergen adalah $k_{eff}$.

```python
import numpy as np
import math
import matplotlib.pyplot as plt
import pandas as pd

# Parameter
jari_jari = 10   #Jari-jari reaktor (cm)
tinggi = 10     #Tinggi reaktor (cm)
jumlah_grup = 5  #Jumlah grup energi
#Jumlah diskritisasi
jumlahtitik_r = 10 
jumlahtitik_z = 10
#Konstanta difusi (cm)
D = [2.5, 1.0, 0.9, 0.8, 0.24]
#Penampang lintang absorpsi (/cm)
sigmaA = [0.04, 0.06, 0.05, 0.17, 3]
#Penampang lintang scattering (/cm)
sigmaS1 = [0.01, 0.015, 0.02, 0.025]   # 1->2, 1->3, 1->4, 1->5
sigmaS2 = [0.005, 0.0075, 0.01]        # 2->3, 2->4, 2->5
sigmaS3 = [0.0025, 0.005]              # 3->4, 3->5
sigmaS4 = [0.001]                      # 4->5
#Nu dan Penampang lintang fisi (/cm)
nu_sigmaF = [0.06, 0.04, 0.05, 0.13, 4.4]
#Chi
chi = [0.9, 0.1, 0, 0, 0]
# nilai keff awal
keff = 1.0
#Diskritisasi ruang
dr = jari_jari / (jumlahtitik_r - 1)
r = np.linspace(0, jari_jari, jumlahtitik_r)
dz = tinggi / (jumlahtitik_z - 1)
z = np.linspace(0, tinggi, jumlahtitik_z)
jumlahtitik = jumlahtitik_r * jumlahtitik_z          # N = Nr*Nz (titik per grup)
matriks_size = jumlahtitik * jumlah_grup             # 5N (ukuran matriks gabungan)

toleransi_k = 1e-10
toleransi_phi = 1e-8
iterasimax = 10000

# Matriks hamburan sigmaS[g, g'] = hamburan dari grup g ke g' (g' > g), lalu removal tiap grup
def matriks_sigmaS(sigmaS1, sigmaS2, sigmaS3, sigmaS4):
    sS = np.zeros((jumlah_grup, jumlah_grup))
    for g, daftar in enumerate([sigmaS1, sigmaS2, sigmaS3, sigmaS4]):
        for n, nilai in enumerate(daftar):
            sS[g, g + 1 + n] = nilai
    return sS

sigmaS = matriks_sigmaS(sigmaS1, sigmaS2, sigmaS3, sigmaS4)
sigmaR = np.array(sigmaA) + sigmaS.sum(axis=1)   # Sigma_R,g = Sigma_a,g + sum_{g'>g} Sigma_s,g->g'

# Penanda titik interior (tidak di r=R, z=0, z=H); titik r=0 termasuk interior (kondisi simetri)
def titik_interior():
    interior = np.zeros(jumlahtitik, dtype=bool)
    for j in range(jumlahtitik_z):
        for i in range(jumlahtitik_r):
            interior[i + j * jumlahtitik_r] = (i < jumlahtitik_r - 1) and (0 < j < jumlahtitik_z - 1)
    return interior

# cek nilai koefisien tiap grup untuk mastiin matriksnya benar (titik interior i=1)
def cek_nilai_grup():
    for g in range(jumlah_grup):
        print(f"Nilai grup {g+1}:")
        print(f"a_P : {2*D[g]/dr**2 + 2*D[g]/dz**2 + sigmaR[g]}")
        print(f"a_W : {D[g]/dr**2 - D[g]/(2*r[1]*dr)}")
        print(f"a_E : {D[g]/dr**2 + D[g]/(2*r[1]*dr)}")
        print(f"a_z : {D[g]/dz**2}")
#cek_nilai_grup()

# Membuat matriks M (satu matriks untuk semua grup, ukuran 5N x 5N)
def matriks_M():
    M = np.zeros((matriks_size, matriks_size))
    for g in range(jumlah_grup):
        for j in range(jumlahtitik_z):
            for i in range(jumlahtitik_r):
                p = i + j * jumlahtitik_r            # indeks titik dalam satu grup
                k = p + g * jumlahtitik              # indeks global
                # Permukaan luar r=R, z=0, z=H : phi = 0
                if i == jumlahtitik_r - 1 or j == 0 or j == jumlahtitik_z - 1:
                    M[k, k] = 1.0
                    continue

                a_z = D[g] / dz**2
                if i == 0:   # sumbu r=0 : simetri (hasil L'Hopital)
                    M[k, k]     = -(4 * D[g] / dr**2 + 2 * a_z + sigmaR[g])
                    M[k, k + 1] = 4 * D[g] / dr**2
                else:        # titik interior
                    a_W = D[g] / dr**2 - D[g] / (2 * r[i] * dr)
                    a_E = D[g] / dr**2 + D[g] / (2 * r[i] * dr)
                    M[k, k]     = -(2 * D[g] / dr**2 + 2 * a_z + sigmaR[g])
                    M[k, k - 1] = a_W
                    M[k, k + 1] = a_E
                # Aksial
                M[k, k - jumlahtitik_r] = a_z
                M[k, k + jumlahtitik_r] = a_z

                # Scattering masuk dari grup lebih tinggi energinya (g' < g) ke grup g
                for gp in range(g):
                    M[k, p + gp * jumlahtitik] = sigmaS[gp, g]

    # Mengecek determinan matriks M (pakai slogdet agar tidak overflow)
    tanda, logdet = np.linalg.slogdet(M)
    if tanda == 0:
        print("Error: matriks M singular, perbaiki nilai masukan.")
    else:
        print(f"Matriks M tidak singular (log|det| = {logdet:.2f})")

    # Cek bentuk akhir matriks M
    pd.DataFrame(M).to_csv("MatriksM.csv", header=False, index=False)
    return M

# Membuat matriks F (satu matriks untuk semua grup): blok (g,g') = chi_g * nuSigmaF_g' * I
def matriks_F():
    F = np.kron(np.outer(chi, nu_sigmaF), np.eye(jumlahtitik))
    F = F * np.tile(titik_interior(), jumlah_grup)[:, None]   # baris titik batas = 0
    pd.DataFrame(F).to_csv("matriksF.csv", header=False, index=False)
    return F

# Iterasi sumber: (-M) phi_baru = F phi / k
def difusi(M, F, keff):
    A_inv = np.linalg.inv(-M)
    phi = np.tile(titik_interior(), jumlah_grup).astype(float)   # tebakan awal
    for n in range(iterasimax):
        sumber = F @ phi
        phi_baru = A_inv @ (sumber / keff)
        keff_baru = keff * np.sum(F @ phi_baru) / np.sum(sumber)
        phi_baru = phi_baru / np.max(phi_baru)                   # normalisasi

        konv_keff = abs(keff_baru - keff) / keff_baru
        konv_phi = np.max(np.abs(phi_baru - phi))
        keff, phi = keff_baru, phi_baru
        if konv_keff < toleransi_k and konv_phi < toleransi_phi:
            print(f"Iterasi selesai pada {n+1}")
            print(f"Nilai keff: {keff:.6f}")
            return phi, keff
    print("Tidak konvergen")
    print(f"Nilai keff : {keff}")
    return phi, keff

# Memanggil semua fungsi itu, huaaaa
M = matriks_M()
F = matriks_F()
phi, keff = difusi(M, F, keff)

abs_phi = abs(phi)
# Membuat grid diskritisasi untuk r dan z
R, Z = np.meshgrid(r, z)

# Merubah ke bentuk 2D untuk masing-masing grup (indeks p = i + j*Nr -> [grup, j(z), i(r)])
phi_2d_grup = abs_phi.reshape((jumlah_grup, jumlahtitik_z, jumlahtitik_r))

# Plot
fig, axes = plt.subplots(2, 3, figsize=(20, 15),constrained_layout=True)
for i, ax in enumerate(axes.flat):
    if i < len(phi_2d_grup): 
        cs = ax.contourf(R, Z, phi_2d_grup[i], levels=50, cmap="viridis")
        fig.colorbar(cs, ax=ax)
        ax.set_xlabel("R (cm)", fontsize=12)
        ax.set_ylabel("Z (cm)", fontsize=12)
        ax.set_title(f"Fluks Neutron Grup {i+1}", fontsize=14)
        ax.set_aspect('auto')
    else:
        ax.axis('off') 
plt.show()
```

## <center>**Distribusi Kerapatan Daya**<center>
Distribusi kerapatan daya dalam teras dapat dihitung dengan:
$$ q^{'''} = E_f \Sigma_f \phi_{r,z} $$
Dimana:
- $q^{'''}$ : kerapatan daya volumetrik $(W/cm^3)$
- $E_f$ : energi yang dilepaskan pada satu reaksi fisi (Joule)
- $\Sigma_f$ : cross section makroskopik fisi (/cm)

```python
# Parameter
Ef = 3.2044e-11 # Energi yang dilepaskan (joule), Setara dengan 200MeV
Sigmaf = [0.06, 0.04, 1.4, 0.13, 0.05] # Penampang lintang fisi (/cm)
q = np.zeros_like(phi) # kerapatan daya
# Mengitung kerapatan daya tiap titik
for i in range(len(phi)):
    grup = i // (jumlahtitik)
    q[i] = phi[i] * Sigmaf[grup] * Ef

# Menghitung presentase kerapatan daya
abs_q = abs(q)
jumlahtotal_q = np.sum(abs_q)
percentage_q = np.array([abs_q[i] / jumlahtotal_q for i in range(len(abs_phi))]) 

# Menghitung daya
totalpowerdensity = 100000/(math.pi*jari_jari**2*tinggi) # W/cc
powerdensity = np.array([percentage_q[i] * totalpowerdensity for i in range(len(percentage_q))])
powerrr = powerdensity.reshape((jumlah_grup, jumlahtitik_r, jumlahtitik_z))

# Menghitung rata-rata power density tiap titik pada 5 grup
rata_rata_grup = np.zeros((jumlah_grup-4, jumlahtitik_r, jumlahtitik_z))
for g in range(jumlah_grup - 4):
    for j in range(jumlahtitik_z):
        for i in range(jumlahtitik_r):
            rata_rata_grup[g, i, j] = np.mean(powerrr[g:g+5, i, j])
rata_rata_grup = rata_rata_grup.reshape((jumlahtitik_r, jumlahtitik_z))
```

## <center>**Transfer panas arah silinder R-Z**</center> 
**Distribusi temperature elemen bahan bakar pada kondisi tunak:**
$$  q^{'''} = -k \nabla^2 T_{(r)}  $$ 

**Laplacian silinder arah (r,z):**
$$ \nabla^2 f_{(r,z)} = \frac{1}{r} \frac{\partial}{\partial r} (r \frac{\partial f_{(r,z)}}{\partial r})+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$
$$ \nabla^2 f_{(r,z)} = \frac{1}{r} (r \frac{\partial}{\partial r} \frac{\partial f_{(r,z)}}{\partial r}+ \frac{\partial r}{\partial r} \frac{\partial f_{(r,z)}}{\partial r})+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$
$$ \nabla^2 f_{(r,z)} = \frac{1}{r} (r \frac{\partial^2 f_{(r,z)}}{\partial r^2}+ 1 \frac{\partial f_{(r,z)}}{\partial r})+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$
$$ \nabla^2 f_{(r,z)} =  \frac{\partial^2 f_{(r,z)}}{\partial r^2}+ \frac{1}{r} \frac{\partial f_{(r,z)}}{\partial r}+ \frac{\partial^2 f_{(r,z)}}{\partial z^2} $$

Diskritisasi arah r, dengan menggunakan central difference
$$ \frac{\partial f_{(r,z)}}{\partial r} = \frac{f_{(r+\Delta r, z)}- f_{(r-\Delta r,z)}}{2 \Delta r} $$
$$ \frac{\partial^2 f_{(r,z)}}{\partial r^2} = \frac{f_{(r+\Delta r, z)}-2 f_{(r,z)}+ f_{(r-\Delta r,z)}}{\Delta r^2} $$
Diskritisasi arah z, dengan menggunakan central difference
$$ \frac{\partial^2 f_{(r,z)}}{\partial z^2} = \frac{f_{(r, z+\Delta z)}-2 f_{(r,z)}+ f_{(r,z-\Delta z)}}{\Delta z^2} $$

Maka setelah di diskritisasi persamaan menjadi
**$$ \nabla^2 f_{(r,z)} = \frac{f_{(r+\Delta r, z)}-2 f_{(r,z)}+ f_{(r-\Delta r,z)}}{\Delta r^2} + \frac{1}{r} \frac{f_{(r+\Delta r, z)}- f_{(r-\Delta r,z)}}{2 \Delta r} + \frac{f_{(r, z+\Delta z)}-2 f_{(r,z)}+ f_{(r,z-\Delta z)}}{\Delta z^2} $$**

**Maka persamaan distribusi temperature menjadi:**
$$ q^{'''} = -k (\frac{T_{(r+\Delta r, z)}-2 T_{(r,z)}+ T_{(r-\Delta r,z)}}{\Delta r^2} + \frac{1}{r} \frac{T_{(r+\Delta r, z)}- T_{(r-\Delta r,z)}}{2 \Delta r} + \frac{T_{(r, z+\Delta z)}-2 T_{(r,z)}+ T_{(r,z-\Delta z)}}{\Delta z^2})  $$ 
$$\frac{q^{'''}}{-k} = \frac{T_{(r+\Delta r, z)}+ T_{(r-\Delta r,z)}}{\Delta r^2} -\frac{2 T_{(r,z)}}{\Delta r^2}+ \frac{1}{r} \frac{T_{(r+\Delta r, z)}- T_{(r-\Delta r,z)}}{2 \Delta r} + \frac{T_{(r, z+\Delta z)}+ T_{(r,z-\Delta z)}}{\Delta z^2} -\frac{2 T_{(r,z)}}{{\Delta z^2}} $$
$$ \frac{q^{'''}}{-k} = \frac{T_{(r+\Delta r, z)}+ T_{(r-\Delta r,z)}}{\Delta r^2} + \frac{1}{r} \frac{T_{(r+\Delta r, z)}- T_{(r-\Delta r,z)}}{2 \Delta r} + \frac{T_{(r, z+\Delta z)}+ T_{(r,z-\Delta z)}}{\Delta z^2} -\frac{T_{(r,z)} (2 \Delta r^2 + 2 \Delta z^2)}{\Delta r^2 \Delta z^2} $$
$$ \frac{T_{(r,z)} (2 \Delta r^2 + 2 \Delta z^2)}{\Delta r^2 \Delta z^2} = \frac{T_{(r+\Delta r, z)}+ T_{(r-\Delta r,z)}}{\Delta r^2} + \frac{1}{r} \frac{T_{(r+\Delta r, z)}- T_{(r-\Delta r,z)}}{2 \Delta r} + \frac{T_{(r, z+\Delta z)}+ T_{(r,z-\Delta z)}}{\Delta z^2} + \frac{q^{'''}}{k} $$
$$ T_{(r,z)} = \frac{\Delta r^2 \Delta z^2}{2 (\Delta r^2 + \Delta z^2)} (\frac{T_{(r+\Delta r, z)}+ T_{(r-\Delta r,z)}}{\Delta r^2} + \frac{1}{r} \frac{T_{(r+\Delta r, z)}- T_{(r-\Delta r,z)}}{2 \Delta r} + \frac{T_{(r, z+\Delta z)}+ T_{(r,z-\Delta z)}}{\Delta z^2} + \frac{q^{'''}}{k}) $$

```python
# Parameter
k = 0.5 # konduktivitas panas

# Membuat matriks T
T = np.zeros((jumlahtitik_r, jumlahtitik_z))
# Boundary condition
T[0, :] = 300
T[:, 0] = 300
T[-1, :] = 300
T[:, -1] = 300

# Distribusi suhu
iterasi = 100
toleransi_T = 1e-3
for x in range(iterasi):
    T_lama = T.copy()
    for j in range(1, jumlahtitik_z - 1):
        for i in range(1, jumlahtitik_r - 1):
            T[i, j] = (((dr**2 * dz**2) / (2 * (dr**2 + dz**2))) * ((T[i + 1, j] + T[i - 1, j]) / dr**2 + 1/jari_jari * ((T[i+1, j] - T[i-1, j])/ 2 * dr) + (T[i, j + 1] + T[i, j - 1]) / dz**2) + rata_rata_grup[i, j] / k)
    konvergensi = np.linalg.norm((T - T_lama) / np.linalg.norm(T + T_lama))
    #print(f"Iterasi ke {i+1}, Nilai konvergensi {konvergensi}")

    if (konvergensi<toleransi_T):
        print(f"Iterasi selesai pada {x+1}")
        print(f"Nilai konvergensi: {konvergensi}")
        break

# Membuat grid diskritisasi untuk r dan z
R, Z = np.meshgrid(r, z)

# Plot
plt.figure(figsize=(16, 8))
plt.subplot(1, 2, 1)
plt.contourf(R, Z, rata_rata_grup, levels=50, cmap="hot")
plt.colorbar()
plt.xlabel("R (cm)")
plt.ylabel("Z (cm)")
plt.title("Distribusi Daya")

plt.subplot(1, 2, 2)
plt.contourf(R, Z, T, levels=50, cmap="hot")
plt.colorbar()
plt.xlabel("R (cm)")
plt.ylabel("Z (cm)")
plt.title("Temperatur fuel")
plt.show()
```
